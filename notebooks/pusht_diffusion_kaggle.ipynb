## 3. Entraînement

Tous les paramètres par défaut de LeRobot pour Diffusion Policy.

Sortie dans `/kaggle/working/outputs/` → accessible dans l'onglet **Output** pour téléchargement.

⚠️ Long run : ~5h sur P100.

```python
!lerobot-train \
  --dataset.repo_id=lerobot/pusht \
  --policy.type=diffusion \
  --policy.push_to_hub=false \
  --env.type=pusht \
  --output_dir=/kaggle/working/outputs/train/diffusion_pusht \
  --job_name=diffusion_pusht \
  --policy.device=cuda
```

## 4. Vérifier les checkpoints produits

```python
!ls -la /kaggle/working/outputs/train/diffusion_pusht/checkpoints/
```

## 5. Évaluation finale (50 épisodes)

Éval automatique faite déjà pendant le training (tous les 20k steps), mais on peut re-évaluer le dernier checkpoint ici.

```python
!lerobot-eval \
  --policy.path=/kaggle/working/outputs/train/diffusion_pusht/checkpoints/last/pretrained_model \
  --env.type=pusht \
  --eval.n_episodes=50 \
  --eval.batch_size=1
```

## 6. Zip du checkpoint final pour téléchargement

Crée une archive dans `/kaggle/working/` — elle apparaîtra dans l'onglet **Output** à droite (clic droit → Download).

```python
import shutil
shutil.make_archive(
    '/kaggle/working/diffusion_pusht_checkpoint',
    'zip',
    '/kaggle/working/outputs/train/diffusion_pusht/checkpoints/last'
)
print('Archive prête : /kaggle/working/diffusion_pusht_checkpoint.zip')
print('→ visible dans l\'onglet Output à droite')
```
